```python
def calculate_metrics(y_true, y_pred, model_name="Model"):
    """
    Calculate RMSE, MAE, and MAPE for model evaluation.
    """
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)

    # MAPE calculation with safety for zero values
    mask = y_true != 0
    mape = np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

    print(f"\n{model_name} Performance:")
    print(f"  RMSE: {rmse:,.2f}")
    print(f"  MAE:  {mae:,.2f}")
    print(f"  MAPE: {mape:.2f}%")

    return {"RMSE": rmse, "MAE": mae, "MAPE": mape}
```

## 5. Основни (Baseline) модели

Основните модели служат како референтна точка за споредба со покомплексни методи.
Во оваа секција се применуваат наивен модел и подвижен просек.


```python
print("\n" + "="*60)
print("📚 PREPARE DATA FOR ALL MODELS")
print("="*60)

test_ts = df.tail(HORIZON).copy()
train_ts = df.iloc[:-HORIZON].copy()

print(f"Train: {train_ts.shape}, Test: {test_ts.shape}")
print(f"Train range: {train_ts.index.min()} to {train_ts.index.max()}")
print(f"Test range: {test_ts.index.min()} to {test_ts.index.max()}")

y_test = test_ts['Passengers'].values
```

```python
print("\n" + "="*60)
print("🧠 BASELINE MODELS")
print("="*60)

# Naive model: last value repeats
naive_pred = np.full(len(test_ts), train_ts['Passengers'].iloc[-1])

# Moving average (12-month window)
ma_12_value = train_ts['Passengers'].rolling(window=12).mean().iloc[-1]
ma_pred = np.full(len(test_ts), ma_12_value)

print(f"\nNaive prediction (last value): {naive_pred[0]:.2f}")
print(f"MA-12 prediction: {ma_pred[0]:.2f}")
```

```python
print("\n" + "="*60)
print("📊 BASELINE MODEL EVALUATION")
print("="*60)

metrics = {}

metrics["Naive"] = calculate_metrics(y_test, naive_pred, "Naive")
metrics["MA-12"] = calculate_metrics(y_test, ma_pred, "Moving Avg (12 months)")
```

## 6. Класични модели (SARIMA и Prophet)


Класичните модели за временски серии експлицитно ја моделираат сезоналноста и трендот.



```python
print("\n" + "="*60)
print("🔮 PROPHET MODEL")
print("="*60)

# Prepare data for Prophet
prophet_df = train_ts.reset_index()
prophet_df.columns = ['ds', 'y']

# Train Prophet
m = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=False,
    daily_seasonality=False
)
m.fit(prophet_df)

# Forecast
future = m.make_future_dataframe(periods=HORIZON, freq='MS')
forecast = m.predict(future)

prophet_pred = forecast['yhat'].iloc[-HORIZON:].values

metrics["Prophet"] = calculate_metrics(y_test, prophet_pred, "Prophet")

# Plot components
m.plot_components(forecast)
plt.show()
```

```python
print("\n🚀 Training SARIMA...")

# SARIMA with monthly seasonality: (p,d,q) x (P,D,Q,12)
sarima_model = SARIMAX(
    train_ts['Passengers'],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12)
)

sarima_fit = sarima_model.fit(disp=False)
sarima_pred = sarima_fit.forecast(steps=HORIZON)

metrics["SARIMA"] = calculate_metrics(y_test, sarima_pred.values, "SARIMA")

print("\n✅ SARIMA training complete")
```

## 7. Машинско учење

Моделите од машинско учење не ја земаат предвид временската зависност директно, па користиме lag features и time features.

### 6.1 Feature Engineering за XGBoost
Создавање на лагови, подвижни просеци и dummy променливи за месеците.


```python
print("\n" + "="*60)
print("🤖 MACHINE LEARNING MODELS")
print("="*60)

def create_lag_features(data, n_lags=12):
    """Create lag features for ML models"""
    df_ml = data.copy()

    # Create lag features (1, 2, 3, 6, 12 months)
    for lag in [1, 2, 3, 6, 12]:
        df_ml[f'lag_{lag}'] = df_ml['Passengers'].shift(lag)

    # Rolling features
    df_ml['roll_mean_3'] = df_ml['Passengers'].shift(1).rolling(window=3).mean()
    df_ml['roll_mean_6'] = df_ml['Passengers'].shift(1).rolling(window=6).mean()
    df_ml['roll_mean_12'] = df_ml['Passengers'].shift(1).rolling(window=12).mean()

    # Time features
    df_ml['month'] = df_ml.index.month
    df_ml['year'] = df_ml.index.year
    df_ml['quarter'] = df_ml.index.quarter

    # Drop NaN from lags
    df_ml = df_ml.dropna()

    return df_ml

# Create features
df_with_lags = create_lag_features(df, N_LAGS)

# Split
train_ml = df_with_lags[df_with_lags.index < test_ts.index.min()]
test_ml = df_with_lags[df_with_lags.index >= test_ts.index.min()]

# Separate features and target
feature_cols = [c for c in df_with_lags.columns if c != 'Passengers']

X_train = train_ml[feature_cols]
y_train_ml = train_ml['Passengers']
X_test = test_ml[feature_cols]
y_test_ml = test_ml['Passengers']

print(f"\nML Train: {X_train.shape}, ML Test: {X_test.shape}")
print(f"Features: {feature_cols}")
```

```python
print("\n🚀 Training Random Forest...")

rf = RandomForestRegressor(
    n_estimators=RF_N_ESTIMATORS,
    max_depth=RF_MAX_DEPTH,
    random_state=RF_RANDOM_STATE,
    n_jobs=-1
)
rf.fit(X_train, y_train_ml)
rf_pred = rf.predict(X_test)

metrics["Random Forest"] = calculate_metrics(y_test_ml, rf_pred, "Random Forest")
```

```python

print("\n🚀 Training XGBoost...")

xgb_model = XGBRegressor(
    n_estimators=XGB_N_ESTIMATORS,
    learning_rate=XGB_LEARNING_RATE,
    max_depth=XGB_MAX_DEPTH,
    random_state=42
)
xgb_model.fit(X_train, y_train_ml)
xgb_pred = xgb_model.predict(X_test)

metrics["XGBoost"] = calculate_metrics(y_test_ml, xgb_pred, "XGBoost")
```

```python

print("\n🚀 Training LightGBM...")

lgbm = LGBMRegressor(
    n_estimators=LGBM_N_ESTIMATORS,
    learning_rate=LGBM_LEARNING_RATE,
    max_depth=LGBM_MAX_DEPTH,
    random_state=42,
    verbose=-1
)
lgbm.fit(X_train, y_train_ml)
lgbm_pred = lgbm.predict(X_test)

metrics["LightGBM"] = calculate_metrics(y_test_ml, lgbm_pred, "LightGBM")
```

```python
model = XGBRegressor(
   n_estimators=600, learning_rate=0.03, max_depth=6, min_child_weight=3, subsample=0.8, colsample_bytree=0.8, gamma=0.1, random_state=42
)

model.fit(X_train, y_train)
```

```python
test['prediction'] = model.predict(X_test)
test.head()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df.index, df['Passengers'], label='Реални вредности')
plt.plot(test.index, test['prediction'], label='XGBoost прогноза')
plt.legend()
plt.xlabel('Време')
plt.ylabel('Број на патници')
plt.title('Time Series прогноза со XGBoost')
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
mae = mean_absolute_error(y_test, test['prediction'])
mse = mean_squared_error(y_test, test['prediction'])
rmse = np.sqrt(mse)
r2 = r2_score(y_test, test['prediction'])

print(f"MAE  (Mean Absolute Error): {mae:.2f}")
print(f"MSE  (Mean Squared Error): {mse:.2f}")
print(f"RMSE (Root Mean Squared Error): {rmse:.2f}")
print(f"R² score: {r2:.3f}")
```
